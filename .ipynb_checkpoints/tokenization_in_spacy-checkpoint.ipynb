```python
import spacy
```

```python
nlp = spacy.blank("eng") # customize based on language you're dealing with. for me is english(en)

doc = nlp("Dr. Emmy loves beer of Rwanda as it cost $2 per bottle.")
```

```python
# word tokenizer
for token in doc:
    print(token)
```
