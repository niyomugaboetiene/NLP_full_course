## Language processing pipeline in spaCy

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
nlp.pipe_names
```

```python
nlp.pipeline
```

```python
doc = nlp("Anna ate $200 of chipps. Then he said  I can do this all day.")

for token in doc:
    print(token, " | ", token.pos_, " | ", token.lemma_) # pos_(part of speech) means the actual meaning of word and lemma_ means the bas verb
```

```python
doc = nlp("Tesla inc is going to buy Twitter $45 billion")
for ent in doc.ents:
    print(ent.text, " | ", ent.label_, " | ", spacy.explain(ent.label_))
```

```python
from spacy import displacy # for displaying or rendering
```

```python
displacy.render(doc, style="ent")
```

```python
# add custome pipeline
source_nlp = spacy.load("en_core_web_sm")

nlp = spacy.blank("en")

nlp.add_pipe("ner", source=source_nlp)
nlp.pipe_names
```
