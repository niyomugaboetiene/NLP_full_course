### regex(regural expression) full course in python

```python
import re
```

```python
chat1 = "Etiene: why you like chips ? my phone number is: 1234567890, and my email: aBc@123.com"
chat2 = "Etiene: here there is (123)-456-7890, abc@xyz.com"
chat3 = "Etiene: yes, your phone number is 1234567894, email: a23c@23a.io"
```

```python
# getting only phone number this several chats
pattern=r'\d{10}|\(\d{3}\)-\d{3}-\d{4}'
matches = re.findall(pattern, chat3)
matches
```

```python
# getting different kind of email address format
pattern2 = r'[a-z0-9A-Z_]*@[a-z0-9A-Z_]*\.[a-zA-Z]*'
matches2 = re.findall(pattern2, chat3)
matches2
```

```python
# select all order id but started by "order" word
chat4 = "Etiene: I'm having issue in my order # 123456789"
chat5 = "Etiene: I have problem with my order number 123456789"
chat6 = "Etiene: My order 123456789 I have issue, I was charget $50 but actual price is $40"
```

```python
pattern3 = r'order[^\d]*(\d*)'
matches3 = re.findall(pattern3, chat4)
matches3
```

```python
text = '''
Queen of the United Kingdom
and other Commonwealth realms (full list)
Reign	6 February 1952 – 8 September 2022
Coronation	2 June 1953
Predecessor	George VI
Successor	Charles III
Born Princess Elizabeth of York
21 April 1926
17 Bruton Street, Mayfair, London, England
Died	8 September 2022 (aged 96)
Balmoral Castle, Aberdeenshire, Scotland
Burial	19 September 2022
King George VI Memorial Chapel, St George's Chapel, Windsor Castle
Spouse	Prince Philip, Duke of Edinburgh
​
​(m. 1947; died 2021)​
Issue	
Charles III
Anne, Princess Royal
Andrew Mountbatten-Windsor
Prince Edward, Duke of Edinburgh
Names Elizabeth Alexandra Mary
House	Windsor
Father	George VI
Mother	Elizabeth Bowes-Lyon
Religion	Protestant[a]'''
```

```python
# extract actual age in the text
pattern4 = r'aged (\d+)'
matches4 = re.findall(pattern4, text)
matches4
```

```python
# extract actual names
# \s capture spaces and .* capture all remaining word of that line
pattern5 = r'Names\s+(.*)'
matches5 = re.findall(pattern5, text)
matches5
```

```python
# extract birth date
pattern6 = r'Born.*\n(.*)'
matches6 = re.findall(pattern6, text)
matches6
```

```python
# extract birth place
pattern7 = r'1926\n(.*)'
matches7 = re.findall(pattern7, text)
matches7
```

```python
def get_pattern_match(pattern, text):
    matches = re.findall(pattern, text)
    if matches:
        return matches[0]
```

```python
get_pattern_match(r'1926\n(.*)', text)
```

```python
def get_personal_information(text):
    age = get_pattern_match(r'aged (\d+)', text)
    name = get_pattern_match(r'Names\s+(.*)', text)
    birth_place = get_pattern_match(r'1926\n(.*)', text)
    birth_date = get_pattern_match(r'Born.*\n(.*)', text)

    return {
        "Age": int(age),
        "Names": name,
        "Birth place": birth_place,
        "Birth date": birth_date
    }
```

```python
get_personal_information(text)
```

### 1. Extract all twitter handles from following text. Twitter handle is the text that appears after https://twitter.com/ and is a single word. Also it contains only alpha numeric characters i.e. A-Z a-z , o to 9 and underscore _

```python
text = '''
Follow our leader Elon musk on twitter here: https://twitter.com/elonmusk, more information 
on Tesla's products can be found at https://www.tesla.com/. Also here are leading influencers 
for tesla related news,
https://twitter.com/teslarati
https://twitter.com/dummy_tesla
https://twitter.com/dummy_2_tesla
'''
pattern = r'https://twitter.com/([a-zA-Z0-9_]+)'

print(re.findall(pattern, text))
```
