## Stemming and Lemmatization with spaCy and NLTK
1. Stemming is just apply fixed rules(ex: eating --> eat, eats --> eat)
2. Lemmatization is just using knowledge on language (ex: ate -> eat)

```python
import spacy
import nltk
```

```python
# stemming
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
```

```python
words = ["eating", "ate", "eats", "adjusting", "adjustable", "running", "coding", "meeting", "ability"]

for word in words:
    print(word, " | ", stemmer.stem(word))
```

```python
#  with spacy (lemmatization)

nlp = spacy.load("en_core_web_sm")

doc = nlp("eating ate eats adjusting adjustable running coding meeting ability better")

for token in doc:
    print(token, " | ", token.lemma_)
```

```python
doc = nlp("Mando talked for 3 hours althrough talking isn't his things he become talkative")

for token in doc:
    print(token.text, " | ", token.lemma_)
```
