## Tokenization with spaCy

```python
import spacy
```

```python
nlp = spacy.blank("eng") # customize based on language you're dealing with. for me is english(en)

doc = nlp("Dr. Emmy loves beer of Rwanda as it cost $2 per bottle.")
```

```python
# word tokenizer
for token in doc:
    print(token)
```

```python
with open("Student_record.txt", "r") as f:
    text = f.readlines()
text
```

```python
text = ' '.join(text)
text
```

```python
# Getting only Email in whole file
doc = nlp(text)
emails = []
for token in doc:
    if token.like_email:
        emails.append(token)
emails
```

```python
doc = nlp("gimme double chees extra large healthy pizza")

tokens = [token.text for token in doc]
tokens
```

```python
from spacy.symbols import ORTH

nlp.tokenizer.add_special_case("gimme", [
    {ORTH: "gim"},
    {ORTH: "me"}
])

doc = nlp("gimme double chees extra large healthy pizza")

tokens = [token.text for token in doc]
tokens
```

```python
nlp.add_pipe("sentencizer")
doc = nlp("Dr. Emmy loves beer of Rwanda as it cost $2 per bottle. Eric likes walking to the street")

for sentences in doc.sents:
    print(sentences)
```
